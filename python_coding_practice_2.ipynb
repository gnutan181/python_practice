```python
def chunk_tokens(
    tokens: list[int],
    chunk_size: int,
    overlap: int
) -> list[dict[int]]:
  result = []
  chunk_id = 0

  for i in range(0, len(tokens), chunk_size - overlap):
    print(i,i + chunk_size)
    # yield tokens[i:i + chunk_size]
    result.append({"chunk_id":chunk_id,"tokens" : tokens[i:i+chunk_size]})
    chunk_id += 1
    # chunks =[]
  return result

tokens = [
    101, 102, 103, 104, 105,
    106, 107, 108, 109, 110
]
chunk_size = 3
overlap = 1
print(list(chunk_tokens(tokens, chunk_size, overlap)))
#
#
```

```python
def two_sum(nums: list[int], target: int) -> list[int]:
    """
    One-pass Hash Table approach.
    Time Complexity: O(N) average, O(N) worst-case (hash collisions)
    Space Complexity: O(N)
    """
    seen: dict[int, int] = {}  # value -> index
    for idx, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], idx]
        seen[num] = idx
    return []


print(two_sum([2,7,9,11],16))
```

Problem 3: Top K Frequent Elements — 3 Evolution StagesStage A: Using collections.Counter ($O(N \log N)$ or $O(N \log K)$)

```python
from collections import Counter

def top_k_frequent_counter(nums: list[int], k: int) -> list[int]:
    # Counter.most_common(k) internally uses heapq.nlargest
    return [item for item , _  in Counter(nums).most_common(k)]



print(top_k_frequent_counter([1,1,2,1,4,2,1,3,3,3,3],2))
```

Optimal Scalable Solution — Min-Heap ($O(N \log K)$ Time, $O(N)$ Space)
For 10 million numbers and $K=10$, sorting all unique elements ($O(U \log U)$ where $U$ is unique elements) wastes massive CPU cycles. Maintaining a Min-Heap of size $K$ guarantees $O(N \log K)$ scaling:

 Move Zeroes

```python
nums = [0, 1, 0, 3, 12,2,0,5,1,0,34,3,0,0,13]

def movezeros(nums):
  k = 0
  for i in range(len(nums)):
    if(nums[i] != 0):
      nums[k], nums[i] = nums[i], nums[k]
      k +=1

  return nums


print(movezeros(nums))
```

 Longest Substring Without Repeating Characters:
 s = "abcabcbb"
Output:
3

```python
def longsubstring(s) -> int:
  i ,j =0 ,0
  charset = set()
  maxlen  = 0
  while j < len(s):

    while s[j] in charset:

      charset.remove(s[i])

      i+=1

    charset.add(s[j])

    maxlen = max(maxlen,j-i+1)

    j +=1
  return maxlen

print(longsubstring("abcadabccb"))
```
