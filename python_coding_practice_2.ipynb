```python
def chunk_tokens(
    tokens: list[int],
    chunk_size: int,
    overlap: int
) -> list[dict[int]]:
  result = []
  chunk_id = 0

  for i in range(0, len(tokens), chunk_size - overlap):
    print(i,i + chunk_size)
    # yield tokens[i:i + chunk_size]
    result.append({"chunk_id":chunk_id,"tokens" : tokens[i:i+chunk_size]})
    chunk_id += 1
    # chunks =[]
  return result

tokens = [
    101, 102, 103, 104, 105,
    106, 107, 108, 109, 110
]
chunk_size = 3
overlap = 1
print(list(chunk_tokens(tokens, chunk_size, overlap)))
#
#
```
