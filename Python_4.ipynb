Write chunk_text(text: str, max_words: int, overlap: int) -> list[str]. Split the text into chunks of at most max_words words. Each chunk should start with the last overlap words of the previous chunk, so context isn’t lost at the boundaries.”

```python
def chunk(text: str, max_words: int, overlap: int) -> list[str]:
    words = text.split()
    # print(len(words), words)
    result = []
    for i in range(0,len(words),max_words-overlap):

      # print(words[i])
      chunk = words[i:i+ max_words-overlap+1]

      # print("ewsd",i,words,chunk)
      result.append(chunk)
    return result      # i += max_words -overlap
```

```python
text = "The cat is brown. She is very naughty. But she loves me. We both watch TV. dfx zdfdx sf dfc dfeddfd "
print(chunk(text,4,1))


# text[1::3]
```

```python
def chunk_text(text: str, max_words: int, overlap: int) -> list[str]:
    """Split text into chunks of at most max_words words, where each chunk
    repeats the last `overlap` words of the previous one."""
    if max_words <= 0:
        raise ValueError("max_words must be positive")
    if not 0 <= overlap < max_words:
        raise ValueError("overlap must be >= 0 and < max_words")  # else infinite loop

    words = text.split()
    print(words)
    if not words:
        return []

    step = max_words - overlap          # how far the window moves each time
    chunks = []
    for start in range(0, len(words), step):
        chunks.append(" ".join(words[start:start + max_words]))
        if start + max_words >= len(words):   # this chunk reached the end
            break
    return chunks
```

```python
text = "The cat is brown. She is very naughty. But she loves me. We both watch TV. ewfdd sdfsdx srdfv fvfd refvd"
print(chunk_text(text,4,1))
```
