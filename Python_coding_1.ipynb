Q1.3 — Group words by first letter.

````python
from collections import defaultdict

def group_by_first_letter(words):
    groups = defaultdict(list)

    for w in words:
        print(w)
        groups[w[0]].append(w)
    print(dict(groups))
    return dict(groups)
````

````python
print(group_by_first_letter(["apple", "banana", "cherry", "date", "elderberry"]))
````

````python
print(group_by_first_letter("szdsfwekdfc"))
````

Q1.4 — Top-k frequent elements.

````python
from collections import Counter

def top_k(items, k):
    print(Counter(items))
    return [item for item, _ in Counter(items).most_common(k)]
````

````python
print(top_k([3,2,1,4,5,1,1,1],2))
````

Q1.5 — Merge two dicts, summing values on key collision.

````python
def merge_dict(d1, d2):
    result = d1  # avoids changing original d1

    for k, v in d2.items():
        if k in result:
            result[k] = result[k] + v
        else:
            result[k] = v

    return result
````

````python
d1={'a': 1, 'b': 2, 'c': 3}
d2={'b': 4, 'c': 5, 's': 6}

print(merge_dict(d1,d2))
print(d1,d2)
````

Q2.4 — map/filter vs comprehension.

````python
nums = [1, 2, 3, 4, 5, 6]

# map/filter style
result = list(map(lambda x: x * x, filter(lambda x: x % 2 == 0, nums)))
print(result)
# comprehension style (generally more idiomatic in Python)
result = [x * x for x in nums if x % 2 == 0]
print(result)
````

Q2.5 — Write a generator for Fibonacci.

````python
def fibonacci(n):
   a,b =0,1
   for i in range(n):
     yield a
     a,b = b, a+b

print(list(fibonacci(6)))
````

Q2.6 — Custom iterator class (the actual protocol, not just yield).

````python
class CountDown():

  def __init__(self,val):
    self.current = val

  def __iter__(self):
    return self

  def __next__(self):
    if self.current <0:
      raise StopIteration
    val =   self.current
    self.current -= 1
    return val
````

````python
counter = CountDown(5)
for i in counter:
  print(i)
````

G2.1 — Late-binding closures in a loop.

Answer: [4, 4, 4, 4, 4], not [0, 1, 2, 3, 4]. Lambdas capture the variable i by reference, not its value at creation time — by the time you call them, the loop has finished and i is 4. Fix: lambda i=i: i (default args are evaluated at definition time, capturing the current value).

````python
funcs = [lambda: i for i in range(5)]
print(funcs)
print([f() for f in funcs])
````
