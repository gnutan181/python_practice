```python
from typing import Callable
def trim_history(messages: list[dict],max_tokens:int,
                 count_tokens : Callable[[str],int] = lambda s : len(s.split()))-> list[dict]:

  """Drop the oldest messages until the history fits max_tokens.
  Always keep the system message(s) and latest message."""

  if not messages:
    return []
  system = [m for m in messages if m["role"] == "system"]
  rest = [m for m in messages if m["role"] != "system"]


  def total(msgs):
    return sum(count_tokens(m["content"]) for m in msgs)


  budget = max_tokens - total(system)
  kept = list(rest)

  while len(kept) > 1 and total(kept) > budget:
      kept.pop(0)

  if total(system + kept) > max_tokens:
    raise ValueError("Latest message is too large to max token")

  return system + kept
```

```python
messages = [
    {"role": "system",    "content": "You are a helpful assistant"},       # 5
    {"role": "user",      "content": "What is RAG"},                       # 3
    {"role": "assistant", "content": "RAG means retrieval augmented generation"},  # 5
    {"role": "user",      "content": "How do I chunk documents"},          # 5
    {"role": "assistant", "content": "Split into overlapping pieces"},     # 4
    {"role": "user",      "content": "What overlap should I use"},         # 5
]
```

```python
def show(title, result):
    print(f"\n=== {title} ===")
    for m in result:
        print(f"  {m['role']:9} | {m['content']}")
    if not result:
        print("  (empty list)")
```

```python
show("Case 1: budget 100, everything fits", trim_history(messages, 100))
# print(type(messages))
```

```python
show("Case 2: budget 20, oldest turns dropped", trim_history(messages, 20))
```

```python
show("Case 3: budget 10, only system + latest", trim_history(messages, 10))
```

```python
def valid_parenthesis(s : str)-> bool:

  stack = []
  pairs = {
      ')' : '(',
      '}':'{',
      ']':'['
  }
  for c in s:
    if c == '(' or c == '{' or c == '[':
      stack.append(c)
    else:
       if len(stack) ==0 :
          return False
       if stack.pop() != pairs[c]:
         return False
  return not stack


print(valid_parenthesis("(){{()}}"))



```
